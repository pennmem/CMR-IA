```python
import pandas as pd
import numpy as np
import pickle
```

```python
from gensim import models
import gensim.downloader as api
```

```python
df = pd.read_csv("simu2b_data/exp1.csv")
df
```

```python
items = np.unique(np.concatenate([df.word1.to_numpy(), df.word2.to_numpy()]))
item2no = {}
for i in range(len(items)):
    item2no[items[i].lower().strip()] = i + 1
wordlist = list(item2no.keys())
wordlist
```

```python
wv = api.load("fasttext-wiki-news-subwords-300")
```

```python
s_mat = np.zeros((len(wordlist), len(wordlist)))
for i in range(len(wordlist)):
    for j in range(len(wordlist)):
        s_mat[i, j] = wv.similarity(wordlist[i], wordlist[j])
s_mat
```

```python
np.save("simu2b_data/simu2b_smat.npy", s_mat)
```

```python
# save item2no
with open("simu2b_data/item2no.pkl", "wb") as f:
    pickle.dump(item2no, f)
```
